```python
import pypowsybl as pp
```

Import load flow module

```python
import pypowsybl.loadflow as lf
from pypowsybl.loadflow import ValidationType
```

Load a test network

```python
n = pp.network.create_four_substations_node_breaker_network()
```

Run loadflow validation

```python
validation_results = lf.run_validation(n, [ValidationType.BUSES, ValidationType.FLOWS, ValidationType.GENERATORS, ValidationType.SVCS, ValidationType.TWTS])
```

Print loadflow validation result

```python
print("Validation :", validation_results.valid)
```

Print a loadflow validation result for buses, branch_flows, generators, svcs and twts

```python
print("Bus :")
print(validation_results.buses)
```

```python
print("Branch Flows :")
print(validation_results.branch_flows)
```

```python
print("Generators :")
print(validation_results.generators)
```

```python
print("SVC :")
print(validation_results.svcs)
```

```python
print("twts :")
print(validation_results.twts)
```
